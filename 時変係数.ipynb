```python
# VSD S系統
import sympy as sp

# 変数を定義
v_d, v_q, K_e, L_d, L_q, omega_re, i_d, i_q, R = sp.symbols('v_d v_q K_e L_d L_q omega_re i_d i_q R')

# 変形したい式を定義
expr =  K_e * ( (v_q - R * i_q - omega_re * L_d * i_d -  2 *omega_re * K_e) / L_q )
expr += 0.5 * (L_d - L_q) * ( (v_d - R * i_d + omega_re * L_q * i_q) / L_d * i_q + i_d * (v_q - R * i_q - omega_re * L_d * i_d - omega_re * 2 * K_e ) / L_q )

# 変数を整理（展開）
expr_expanded = sp.expand(expr)

# v_d, v_q に関する項を抽出
A = sp.collect(expr_expanded, v_d, evaluate=False).get(v_d, 0)  # v_d の係数
B = sp.collect(expr_expanded, v_q, evaluate=False).get(v_q, 0)  # v_q の係数
C = expr_expanded.subs({v_d: 0, v_q: 0})  # 定数項

# 変形後の式を表示
transformed_expr = A * v_d + B * v_q + C
#print("変形後の式:", transformed_expr)
print(f"A: {A}")
print(f"B: {B}")
print(f"C: {C}")

#sp.simplify(A)
#sp.simplify(B)
sp.simplify(C)
```
